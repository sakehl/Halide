# Experiments for HaliVer

```python
from preprocess import *
from typing import List, Optional, Dict, Tuple, Any
import subprocess
import os
import time
import json
```

```python
def runCommand(command: str, verbose: bool = False)-> Tuple[float, bool]:
    start = time.time()
    p = subprocess.run(command, stdout=subprocess.PIPE, stderr=subprocess.PIPE, shell=True)
    end = time.time()
    if(p.returncode != 0):
        print(p.stdout.decode('UTF-8'))
        print(p.stderr.decode('UTF-8'))
        return end - start, False
    elif verbose:
        print(p.stdout.decode('UTF-8'))
        
    return end - start, True
```

```python
compilationTimes: Dict[str, float] = {}
results: Dict[str, Any] = {}
verificationTimes: Dict[str, Tuple[float, bool]] = {}
```

## Blur

```python
name = 'blur'
pvl_name = 'build/' + name + '_pvl.pvl'
```

```python
compileTimeStage1, _ = runCommand('cmake --build build -t ' + name + ' --clean-first')
print(compileTimeStage1)
```

```python
# Front-end verification
full_name = name + '-' + 'front'
compileTimeStage2 = runCommand('cd build && ./' + name + ' ' + 'front')[0]
compilationTimes[full_name] = compileTimeStage1 + compileTimeStage2

pvl_name = name + '_front_pvl.pvl'
verificationTime = runCommand('vct build/' + name + '_front_pvl.pvl')
verificationTimes[full_name] = verificationTime
print(verificationTime)
```

```python
# Back-end verification
for v in ['0', '1', '2', '3']:
    full_name = name + '-' + v
    compileTimeStage2 = runCommand('cd build && ./' + name + ' ' + v)[0]
    compilationTimes[full_name] = compileTimeStage1 + compileTimeStage2
    line_info = count_lines(name, v)
    results[full_name] = line_info.__dict__
    print(line_info)
```

```python
for v in ['0', '1', '2', '3']:
    full_name = name + '-' + v
    pvl_name = name + '_pvl-' + v + '.pvl'
    verificationTime = runCommand('vct build/' + pvl_name)
    verificationTimes[full_name] = verificationTime
    print(verificationTime)
```

```python
with open(name + '_results.txt', 'w') as f:
    f.write(json.dumps(compilationTimes))
    f.write("\n")
    f.write(json.dumps(results))
    f.write("\n")
    f.write(json.dumps(verificationTimes))
```
